#  Import Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

# Dataset

```python
data = {
    'text': [
        "Win a brand new car",
        "Hello how are you",
        "Please send the report",
        "Free gift voucher available",
        "Let's meet tomorrow",
        "Click here to win prize",
        "I will call you tonight",
        "Exclusive deal just for you",
        "Are you coming to class",
        "Claim your reward now"
    ],

    'label': [
        1, 0, 0, 1, 0,
        1, 0, 1, 0, 1
    ]
}
df=pd.DataFrame(data)
print(df)
```

# Feature Extraction (Bag-of-Words)

```python
vectorize = CountVectorizer()
x=vectorize.fit_transform(df['text'])
y=df['label']
```

# Data Split

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

# Train Naive Bayes Model

```python
model = MultinomialNB()
model.fit(x_train, y_train)
```

# Model Testing

```python
y_pred=model.predict(x_test)
print("Predictions:",y_pred)
print("Actual:",y_test.values)
```

# Model Evaluation

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:", classification_report(y_test, y_pred))
```

# Model Saving

```python
import joblib

joblib.dump(model, 'naive_bayes_model.pkl')
joblib.dump(vectorize, 'vectorizer.pkl')
```

# Load Model

```python
loaded_model = joblib.load('naive_bayes_model.pkl')
loaded_vectorizer = joblib.load('vectorizer.pkl')
```

# Model Deployment

```python
text = ["Free offer now"]
text_vec = loaded_vectorizer.transform(text)
prediction = loaded_model.predict(text_vec)
print("Prediction:", prediction[0])
```

# Case study- 2:NEWS DATA

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

# LOAD DATASET


```python
categories = ['sci.space', 'rec.sport.hockey']
data = fetch_20newsgroups(subset='all', categories=categories)

texts = data.data
labels = data.target
for i in range(3):
    print("Text:", texts[i][:200])
    print("Label:", labels[i])
    print("------")
```

# Text processing

```python
import re
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text
texts = [preprocess(t) for t in texts]
```

# Feature Extraction

```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
```

# data split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)
```

# Model training

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

# Model Testing

```python
y_pred = model.predict(X_test)
print("Predictions:", y_pred)
print("Actual:", y_test)
```

# Model Evaluation

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

# MODEL SAVING

```python
import joblib
joblib.dump(model, 'naive_bayes_model_news.pkl')
joblib.dump(vectorizer, 'vectorizer_news.pkl')
```

# LOAD MODEL

```python
loaded_model = joblib.load('naive_bayes_model_news.pkl')
loaded_vectorizer = joblib.load('vectorizer_news.pkl')
```

Model deployment

```python
test_samples = [
    "The football team secured a thrilling victory in the final match",
    "Scientists discovered a new exoplanet in a distant galaxy",
    "The basketball league announced its schedule for the upcoming season"
]
test_vec = loaded_vectorizer.transform(test_samples)
predictions = loaded_model.predict(test_vec)
for text, pred in zip(test_samples, predictions):
    print("\nText:", text)
    print("Predicted Category:", data.target_names[pred])
```
