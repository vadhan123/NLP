```python
# Define the text directly in a variable
text = """Natural Language Processing (NLP) is a branch of artificial intelligence that focuses on enabling computers to understand, interpret, and generate human language in a meaningful way. It combines techniques from computer science, linguistics, and machine learning to analyze text and speech. NLP is used in many everyday applications such as chatbots, voice assistants, language translation, sentiment analysis, and spell checking. By allowing machines to process natural language, NLP helps bridge the gap between human communication and computer understanding, making technology more interactive and user-friendly."""

# Perform simple preprocessing tasks

# 1. Count the number of words in the text
# We'll use split() to separate words based on whitespace
if text:
    words = text.split()
    word_count = len(words)
    print(f"\nNumber of words: {word_count}")
else:
    print("\nNo text to process for word count.")

# 2. Convert the entire text to lowercase
if text:
    lowercase_text = text.lower()
    print("\nText in Lowercase:")
    print(lowercase_text)
else:
    print("\nNo text to convert to lowercase.")
```
